```python
import tensorflow as tf

# Define the paths to the training and validation data directories
train_dir = '/content/Dataset_extracted/Train'
test_dir = '/content/Dataset_extracted/Validation'


# Load the training dataset from the directory structure
train_data = tf.keras.utils.image_dataset_from_directory(
    train_dir,
    labels='inferred',          # Automatically generate labels based on the names of the subdirectories
    label_mode='categorical',   # Convert labels to one-hot encoded vectors (for multi-class classification)
    batch_size=64,              # Group the data into batches of 64 images
    image_size=(128, 128),      # Resize all input images to a uniform size of 128x128 pixels
    shuffle=True                # Randomly shuffle the dataset at the start of each epoch to ensure variety
)


# Load the validation dataset from the directory structure
validation_data = tf.keras.utils.image_dataset_from_directory(
    test_dir,
    labels='inferred',          # Automatically generate labels based on the names of the subdirectories
    label_mode='categorical',   # Convert labels to one-hot encoded vectors to match the training data
    batch_size=64,              # Group the data into batches of 64 images
    image_size=(128, 128),      # Resize all input images to a uniform size of 128x128 pixels
    shuffle=False               # Keep the data order sequential for consistent evaluation metrics
)


# Set up AUTOTUNE to dynamically adjust CPU processing pipeline performance during data loading
AUTOTUNE = tf.data.AUTOTUNE
```

```python
from tensorflow.keras import layers, models

# Create model structure
model = models.Sequential([
    layers.Input(shape=(128, 128, 3)),

    # Feature extraction layers
    layers.Conv2D(32, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),

    # Classification layers
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dense(2, activation='softmax')
])

# Compile model
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# Show model summary
model.summary()
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, BatchNormalization, Flatten, Dense, Dropout

# Build the CNN model
model = Sequential()


model.add(Conv2D(32, kernel_size=(3,3), padding='valid', activation='relu', input_shape=(128, 128, 3)))
model.add(BatchNormalization())
model.add(MaxPooling2D(pool_size=(2,2), strides=2, padding='valid'))


model.add(Conv2D(64, kernel_size=(3,3), padding='valid', activation='relu'))
model.add(BatchNormalization())
model.add(MaxPooling2D(pool_size=(2,2), strides=2, padding='valid'))


model.add(Conv2D(128, kernel_size=(3,3), padding='valid', activation='relu'))
model.add(BatchNormalization())
model.add(MaxPooling2D(pool_size=(2,2), strides=2, padding='valid'))


model.add(Flatten())

# Fully Connected Layers (Classifier)
model.add(Dense(128, activation='relu'))
model.add(Dropout(0.1))

model.add(Dense(64, activation='relu'))
model.add(Dropout(0.1))


model.add(Dense(2, activation='softmax'))

# Compile the model with correct loss function
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',  # Matches categorical labels used today
    metrics=['accuracy']
)

# Display model summary
model.summary()
```

```python
import tensorflow as tf

# Define data paths
train_dir = '/content/Dataset_extracted/Train'
validation_dir = '/content/Dataset_extracted/Validation'
test_dir = '/content/Dataset_extracted/Test'


# Load training dataset
train_data = tf.keras.utils.image_dataset_from_directory(
    train_dir,
    label_mode='categorical',
    batch_size=16,
    image_size=(32, 32),
    shuffle=True
)

# Load validation dataset
validation_data = tf.keras.utils.image_dataset_from_directory(
    validation_dir,
    label_mode='categorical',
    batch_size=16,
    image_size=(32, 32),
    shuffle=False
)

# Load test dataset
test_data = tf.keras.utils.image_dataset_from_directory(
    test_dir,
    label_mode='categorical',
    batch_size=16,
    image_size=(32, 32),
    shuffle=False
)


# Extract true labels from the test pipeline
y_true = tf.concat([y for x, y in test_data], axis=0).numpy()
y_true_labels = tf.argmax(y_true, axis=1).numpy()

```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten

# Initialize a sequential container for the custom ResNet model
resnet_model = Sequential()

# Load the core ResNet50 architecture pre-trained on ImageNet without its top classification layer
pretrained_model = tf.keras.applications.ResNet50(
    include_top=False,
    input_shape=(32, 32, 3),
    pooling='max',
    classes=2,
    weights='imagenet'
)

# Freeze all layers of the pre-trained ResNet50 model to keep weights unchanged
for layer in pretrained_model.layers:
    layer.trainable = False

# Construct the final network layout
resnet_model.add(pretrained_model)
resnet_model.add(Flatten())
resnet_model.add(Dense(128, activation='relu'))
resnet_model.add(Dense(2, activation='sigmoid'))

```

```python
# Compile the custom ResNet50 model
resnet_model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
# Train the ResNet50 model for 5 epochs
history = resnet_model.fit(
    train_data,
    epochs=5,
    validation_data=validation_data
)
```
